```python

# ============================================================
# FRANCK-HERTZ | DETECCIÓN DE MÍNIMOS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import find_peaks
from scipy.stats import linregress

LIMITE = 12.67
PROMINENCIA = 0.8
DISTANCIA = 3


# ------------------------------------------------------------
# 1. SEPARAR LA SECUENCIA EN TRAMOS VÁLIDOS
# ------------------------------------------------------------

def obtener_tramos(U, I):

    n = min(len(U), len(I))
    U, I = np.asarray(U[:n], float), np.asarray(I[:n], float)

    validos = np.isfinite(U) & np.isfinite(I) & (I < LIMITE)

    tramos = []
    actual = []

    for i in range(n):

        if not validos[i]:
            if len(actual) >= 3:
                tramos.append(actual)
            actual = []
            continue

        # Un retroceso rompe la continuidad del barrido
        if actual and U[i] <= U[actual[-1]]:
            if len(actual) >= 3:
                tramos.append(actual)
            actual = []

        actual.append(i)

    if len(actual) >= 3:
        tramos.append(actual)

    return U, I, tramos


# ------------------------------------------------------------
# 2. DETECTAR MÍNIMOS EN CADA TRAMO
# ------------------------------------------------------------

def analizar_minimos(U, I, numero):

    U, I, tramos = obtener_tramos(U, I)
    candidatos = []

    for numero_tramo, indices in enumerate(tramos, start=1):

        corriente = I[indices]

        posiciones, propiedades = find_peaks(
            -corriente,
            prominence=PROMINENCIA,
            distance=DISTANCIA
        )

        for j, posicion in enumerate(posiciones):

            indice = indices[posicion]

            candidatos.append({
                "Toma": numero,
                "Tramo": numero_tramo,
                "Registro": indice + 1,
                "U2 (V)": U[indice],
                "I (nA)": I[indice],
                "Prominencia (nA)": propiedades["prominences"][j]
            })

    resultados = pd.DataFrame(candidatos)

    print(f"\n{'='*55}")
    print(f"MÍNIMOS CANDIDATOS · TOMA {numero}")
    print(f"{'='*55}")

    if resultados.empty:
        print("No se encontraron mínimos candidatos.")
        return resultados

    resultados = resultados.sort_values("Registro").reset_index(drop=True)

    print(resultados.round(3).to_string(index=False))

    # --------------------------------------------------------
    # 3. CALCULAR SEPARACIONES ENTRE MÍNIMOS
    # --------------------------------------------------------

    # Solo se comparan mínimos consecutivos del mismo tramo.
    separaciones = []

    for tramo, grupo in resultados.groupby("Tramo"):

        voltajes = grupo["U2 (V)"].to_numpy()

        for j, delta in enumerate(np.diff(voltajes)):

            separaciones.append({
                "Tramo": tramo,
                "Intervalo": f"{j+1} - {j+2}",
                "Separación (V)": delta
            })

    tabla_sep = pd.DataFrame(separaciones)

    if not tabla_sep.empty:

        print("\nSEPARACIONES ENTRE MÍNIMOS")
        print(tabla_sep.round(3).to_string(index=False))

        promedio = tabla_sep["Separación (V)"].mean()

        print(f"\nPromedio provisional: {promedio:.3f} V")
        print(f"Energía equivalente: {promedio:.3f} eV")

    # --------------------------------------------------------
    # 4. AJUSTE LINEAL POR TRAMO
    # --------------------------------------------------------

    for tramo, grupo in resultados.groupby("Tramo"):

        if len(grupo) < 3:
            continue

        voltajes = grupo["U2 (V)"].to_numpy()
        orden = np.arange(1, len(voltajes) + 1)

        ajuste = linregress(orden, voltajes)

        print(f"\nAJUSTE LINEAL · TRAMO {tramo}")
        print(f"Pendiente: {ajuste.slope:.3f} V")
        print(f"Intercepto: {ajuste.intercept:.3f} V")
        print(f"R²: {ajuste.rvalue**2:.4f}")
        print(
            f"Energía equivalente: "
            f"{ajuste.slope:.3f} eV"
        )

    # --------------------------------------------------------
    # 5. GRÁFICA DE LOS MÍNIMOS
    # --------------------------------------------------------

    fig, ax = plt.subplots(figsize=(9, 5.5))

    validos = np.isfinite(U) & np.isfinite(I) & (I < LIMITE)

    ax.scatter(
        U[validos], I[validos],
        s=30, color="#245B78",
        edgecolor="white", linewidth=0.4,
        label="Mediciones válidas"
    )

    if not resultados.empty:

        ax.scatter(
            resultados["U2 (V)"],
            resultados["I (nA)"],
            marker="v", s=90,
            color="#D58B35",
            edgecolor="white",
            label="Mínimos candidatos",
            zorder=4
        )

    ax.set(
        title=f"Mínimos locales · Toma {numero}",
        xlabel=r"Potencial acelerador $U_2$ (V)",
        ylabel=r"Corriente colectada $I_c$ (nA)"
    )

    ax.grid(True)
    ax.legend(frameon=False)
    fig.tight_layout()

    fig.savefig(
        f"minimos_toma_{numero}.png",
        dpi=300, bbox_inches="tight"
    )

    plt.show()

    return resultados


# ------------------------------------------------------------
# 6. EJECUTAR PARA LAS DOS TOMAS
# ------------------------------------------------------------

minimos_1 = analizar_minimos(U2_1, I_1, 1)
minimos_2 = analizar_minimos(U2_2, I_2, 2)


# ------------------------------------------------------------
# 7. EXPORTAR RESULTADOS
# ------------------------------------------------------------

minimos_1.to_csv("minimos_toma_1.csv", index=False)
minimos_2.to_csv("minimos_toma_2.csv", index=False)

print("\nAnálisis de mínimos finalizado.")
```

```python

import numpy as np
import pandas as pd

LIMITE_MEDIDOR = 12.67

def auditar_toma(U, I, numero):
    U = np.asarray(U, dtype=float)
    I = np.asarray(I, dtype=float)

    n = max(len(U), len(I))
    filas = []

    for k in range(n):
        u = U[k] if k < len(U) else np.nan
        corriente = I[k] if k < len(I) else np.nan

        observaciones = []

        if np.isnan(u):
            observaciones.append("Falta voltaje")
        if np.isnan(corriente):
            observaciones.append("Falta corriente")

        if np.isfinite(corriente):
            if corriente >= LIMITE_MEDIDOR:
                observaciones.append("Saturación o sobreescala")
            elif corriente < 0:
                observaciones.append("Corriente negativa: conservar")

        if np.isfinite(u) and k > 0 and k < len(U):
            if u < U[k-1]:
                observaciones.append("Retroceso en voltaje")

            if u - U[k-1] > 3:
                observaciones.append("Salto de voltaje mayor a 3 V")

        if np.isfinite(corriente) and corriente > LIMITE_MEDIDOR:
            observaciones.append("Lectura superior al límite instrumental")

        estado = "Válido"
        if any("Falta" in x for x in observaciones):
            estado = "Incompleto"
        elif any("Saturación" in x or "superior" in x
                 for x in observaciones):
            estado = "Saturado"
        elif any("Retroceso" in x or "Salto" in x
                 for x in observaciones):
            estado = "Revisar"

        filas.append({
            "Registro": k + 1,
            "U2_V": u,
            "I_nA": corriente,
            "Estado": estado,
            "Observaciones": "; ".join(observaciones)
        })

    tabla = pd.DataFrame(filas)

    # Se conserva cada valor original y se exporta también
    # una tabla con las posiciones que no tienen pareja.
    tabla.to_csv(
        f"auditoria_toma_{numero}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n{'='*45}")
    print(f"AUDITORÍA DE LA TOMA {numero}")
    print(f"{'='*45}")
    print(f"Voltajes originales: {len(U)}")
    print(f"Corrientes originales: {len(I)}")
    print(f"Filas de la tabla: {n}")
    print(f"Registros emparejados por posición: {min(len(U),len(I))}")
    print(f"Voltajes sin corriente: {max(0,len(U)-len(I))}")
    print(f"Corrientes sin voltaje: {max(0,len(I)-len(U))}")

    print("\nResumen:")
    print(tabla["Estado"].value_counts())

    print("\nRegistros que requieren revisión:")
    print(tabla[tabla["Estado"] != "Válido"].to_string(index=False))

    return tabla


tabla_1 = auditar_toma(U2_1, I_1, 1)
tabla_2 = auditar_toma(U2_2, I_2, 2)

print("\nAuditoría terminada.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

# --------------------------------------------------
# CONFIGURACIÓN
# --------------------------------------------------

LIMITE = 12.67

# Umbrales para estudiar la estabilidad de los mínimos
PROMINENCIAS = [0.4, 0.6, 0.8, 1.0]

# Separación mínima entre candidatos, expresada en muestras
DISTANCIA = 3

PROMINENCIA_PRINCIPAL = 0.8


# --------------------------------------------------
# 1. SEPARAR LOS DATOS EN TRAMOS VÁLIDOS
# --------------------------------------------------

def obtener_tramos(tabla):

    U = tabla["U2_V"].to_numpy(dtype=float)
    I = tabla["I_nA"].to_numpy(dtype=float)
    registros = tabla["Registro"].to_numpy()

    completos = np.isfinite(U) & np.isfinite(I)
    validos = completos & (I < LIMITE)

    tramos = []
    actual = []

    for i in range(len(U)):

        if not validos[i]:

            if len(actual) >= 3:
                tramos.append(actual)

            actual = []
            continue

        # Si el voltaje retrocede o se repite,
        # comenzamos un nuevo tramo.
        if actual and U[i] <= U[actual[-1]]:

            if len(actual) >= 3:
                tramos.append(actual)

            actual = []

        actual.append(i)

    if len(actual) >= 3:
        tramos.append(actual)

    return U, I, registros, tramos


# --------------------------------------------------
# 2. DETECTAR LOS MÍNIMOS LOCALES
# --------------------------------------------------

def detectar_minimos(tabla, toma):

    U, I, registros, tramos = obtener_tramos(tabla)

    resultados = []

    print(f"\n{'='*55}")
    print(f"ANÁLISIS DE MÍNIMOS — TOMA {toma}")
    print(f"{'='*55}")

    for prominencia in PROMINENCIAS:

        candidatos = []

        for numero_tramo, indices in enumerate(tramos, start=1):

            corriente = I[indices]

            posiciones, propiedades = find_peaks(
                -corriente,
                prominence=prominencia,
                distance=DISTANCIA
            )

            for j, posicion in enumerate(posiciones):

                indice = indices[posicion]

                candidatos.append({
                    "Toma": toma,
                    "Tramo": numero_tramo,
                    "Registro": registros[indice],
                    "U2_V": U[indice],
                    "I_nA": I[indice],
                    "Prominencia_nA": propiedades["prominences"][j],
                    "Umbral_nA": prominencia,
                    "I_anterior_nA": I[indices[posicion - 1]],
                    "I_posterior_nA": I[indices[posicion + 1]]
                })

        resultados.extend(candidatos)

        print(
            f"Prominencia mínima {prominencia:.1f} nA: "
            f"{len(candidatos)} candidatos"
        )

    tabla_completa = pd.DataFrame(resultados)

    if tabla_completa.empty:
        print("No se detectaron mínimos.")
        return tabla_completa

    # Exportar todos los candidatos para comparar umbrales.
    tabla_completa.to_csv(
        f"candidatos_minimos_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    # Seleccionar los candidatos del umbral principal.
    seleccion = tabla_completa[
        np.isclose(
            tabla_completa["Umbral_nA"],
            PROMINENCIA_PRINCIPAL
        )
    ].copy()

    seleccion.to_csv(
        f"minimos_principales_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("\nCandidatos con prominencia principal:")
    print(
        seleccion[
            ["Tramo", "Registro", "U2_V", "I_nA", "Prominencia_nA"]
        ].to_string(index=False)
    )

    # --------------------------------------------------
    # 3. GRAFICAR LOS MÍNIMOS CANDIDATOS
    # --------------------------------------------------

    fig, ax = plt.subplots(figsize=(10, 5.8))

    for numero_tramo, indices in enumerate(tramos, start=1):

        ax.scatter(
            U[indices],
            I[indices],
            s=32,
            alpha=0.75,
            label=f"Tramo {numero_tramo}"
        )

    if not seleccion.empty:

        ax.scatter(
            seleccion["U2_V"],
            seleccion["I_nA"],
            marker="v",
            s=110,
            color="#D9534F",
            edgecolor="black",
            linewidth=0.6,
            label=f"Mínimos candidatos ({PROMINENCIA_PRINCIPAL} nA)",
            zorder=5
        )

    ax.set_title(f"Mínimos locales de corriente — Toma {toma}")
    ax.set_xlabel(r"Voltaje acelerador $U_2$ (V)")
    ax.set_ylabel("Corriente colectora (nA)")

    ax.grid(alpha=0.2, linestyle=":")
    ax.legend(frameon=False, fontsize=9)

    plt.tight_layout()

    plt.savefig(
        f"minimos_candidatos_toma_{toma}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    return tabla_completa


# --------------------------------------------------
# 4. EJECUTAR PARA AMBAS TOMAS
# --------------------------------------------------

candidatos_1 = detectar_minimos(tabla_1, 1)
candidatos_2 = detectar_minimos(tabla_2, 2)

print("\nAnálisis de detección terminado.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# --------------------------------------------------
# CONFIGURACIÓN
# --------------------------------------------------

ENERGIA_REFERENCIA = 18.7  # eV, valor aproximado para comparación

plt.rcParams.update({
    "figure.figsize": (9, 5.5),
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "grid.alpha": 0.2
})


# --------------------------------------------------
# 1. SELECCIONAR LOS MÍNIMOS PRINCIPALES
# --------------------------------------------------

def seleccionar_minimos(candidatos, toma):

    if candidatos.empty:
        print(f"No hay candidatos para la toma {toma}.")
        return pd.DataFrame()

    seleccion = candidatos[
        np.isclose(candidatos["Umbral_nA"], 0.8)
    ].copy()

    seleccion = seleccion.sort_values(
        ["Tramo", "U2_V"]
    ).reset_index(drop=True)

    return seleccion


# --------------------------------------------------
# 2. ANALIZAR LAS SEPARACIONES Y AJUSTES
# --------------------------------------------------

def analizar_energia(candidatos, toma):

    minimos = seleccionar_minimos(candidatos, toma)

    if minimos.empty:
        return pd.DataFrame(), pd.DataFrame()

    separaciones_totales = []
    ajustes = []

    print("\n" + "=" * 60)
    print(f"ANÁLISIS DE ENERGÍA — TOMA {toma}")
    print("=" * 60)

    for tramo, grupo in minimos.groupby("Tramo"):

        grupo = grupo.sort_values("U2_V").reset_index(drop=True)

        voltajes = grupo["U2_V"].to_numpy()
        numero_minimos = len(voltajes)

        print(f"\nTRAMO {tramo}")
        print(f"Número de mínimos: {numero_minimos}")

        if numero_minimos < 2:
            print("No hay suficientes mínimos para calcular separaciones.")
            continue

        # Diferencias entre mínimos consecutivos
        separaciones = np.diff(voltajes)

        print("\nSeparaciones entre mínimos:")
        for i, delta in enumerate(separaciones, start=1):
            print(
                f"ΔU_{i} = {voltajes[i]:.3f} - "
                f"{voltajes[i-1]:.3f} = {delta:.3f} V"
            )

            separaciones_totales.append({
                "Toma": toma,
                "Tramo": tramo,
                "Separacion": i,
                "Delta_U_V": delta
            })

        promedio = np.mean(separaciones)

        desviacion = (
            np.std(separaciones, ddof=1)
            if len(separaciones) > 1 else np.nan
        )

        error_estandar = (
            desviacion / np.sqrt(len(separaciones))
            if len(separaciones) > 1 else np.nan
        )

        print(f"\nSeparación media: {promedio:.3f} V")

        if np.isfinite(desviacion):
            print(f"Desviación estándar: {desviacion:.3f} V")
            print(f"Error estándar: {error_estandar:.3f} V")
        else:
            print("Solo existe una separación: no se puede estimar "
                  "la dispersión entre separaciones.")

        # --------------------------------------------------
        # AJUSTE LINEAL
        # U_n = a + b*n
        # --------------------------------------------------

        if numero_minimos >= 3:

            orden = np.arange(numero_minimos)

            ajuste = linregress(orden, voltajes)

            pendiente = ajuste.slope
            intercepto = ajuste.intercept

            energia = pendiente  # eV

            discrepancia = (
                abs(energia - ENERGIA_REFERENCIA)
                / ENERGIA_REFERENCIA * 100
            )

            print("\nAJUSTE LINEAL")
            print(f"U_n = {intercepto:.3f} + "
                  f"{pendiente:.3f} n")
            print(f"R² = {ajuste.rvalue**2:.5f}")
            print(f"Error estándar de la pendiente: "
                  f"{ajuste.stderr:.3f} V")

            print(f"\nEnergía estimada: {energia:.3f} eV")
            print(f"Referencia: {ENERGIA_REFERENCIA:.2f} eV")
            print(f"Discrepancia: {discrepancia:.2f} %")

            ajustes.append({
                "Toma": toma,
                "Tramo": tramo,
                "Numero_minimos": numero_minimos,
                "Intercepto_V": intercepto,
                "Energia_eV": energia,
                "Error_pendiente_eV": ajuste.stderr,
                "R2": ajuste.rvalue**2,
                "Discrepancia_porcentual": discrepancia
            })

            # Gráfica del ajuste
            fig, ax = plt.subplots(figsize=(8, 5))

            ax.scatter(
                orden,
                voltajes,
                color="#2864A5",
                s=65,
                label="Mínimos experimentales",
                zorder=3
            )

            ax.plot(
                orden,
                ajuste.intercept + ajuste.slope * orden,
                color="#D9534F",
                linewidth=1.8,
                label="Ajuste lineal"
            )

            ax.set_title(
                f"Ajuste lineal — Toma {toma}, tramo {tramo}"
            )
            ax.set_xlabel("Orden del mínimo")
            ax.set_ylabel(r"Voltaje $U_2$ (V)")
            ax.grid(alpha=0.2, linestyle=":")
            ax.legend(frameon=False)

            plt.tight_layout()

            plt.savefig(
                f"ajuste_lineal_toma_{toma}_tramo_{tramo}.png",
                bbox_inches="tight"
            )

            plt.show()

        else:
            print(
                "\nNo se realiza ajuste lineal: "
                "se requieren al menos tres mínimos."
            )

    # --------------------------------------------------
    # TABLAS FINALES
    # --------------------------------------------------

    tabla_separaciones = pd.DataFrame(separaciones_totales)
    tabla_ajustes = pd.DataFrame(ajustes)

    tabla_separaciones.to_csv(
        f"separaciones_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    tabla_ajustes.to_csv(
        f"energia_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    return tabla_separaciones, tabla_ajustes


# --------------------------------------------------
# 3. EJECUTAR PARA AMBAS TOMAS
# --------------------------------------------------

separaciones_1, energia_1 = analizar_energia(candidatos_1, 1)

separaciones_2, energia_2 = analizar_energia(candidatos_2, 2)

print("\nANÁLISIS COMPLETADO.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# ==================================================
# CONFIGURACIÓN
# ==================================================

ENERGIA_REFERENCIA = 18.7
REFERENCIA_MIN = 18.4
REFERENCIA_MAX = 19.0

PROMINENCIA_SELECCIONADA = 0.8

# No introducir una incertidumbre instrumental
# sin conocer la resolución real del voltímetro.

plt.rcParams.update({
    "figure.figsize": (9, 5.5),
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "grid.alpha": 0.20
})


# ==================================================
# 1. ANÁLISIS ESTADÍSTICO DE CADA TRAMO
# ==================================================

def analizar_estadistica(candidatos, toma):

    resultados = []
    separaciones_totales = []

    if candidatos.empty:
        print(f"No hay candidatos para la toma {toma}.")
        return pd.DataFrame(), pd.DataFrame()

    minimos = candidatos[
        np.isclose(
            candidatos["Umbral_nA"],
            PROMINENCIA_SELECCIONADA
        )
    ].copy()

    if minimos.empty:
        print(f"No hay mínimos seleccionados en la toma {toma}.")
        return pd.DataFrame(), pd.DataFrame()

    for tramo, grupo in minimos.groupby("Tramo"):

        grupo = grupo.sort_values("U2_V")

        voltajes = grupo["U2_V"].to_numpy(dtype=float)
        voltajes = voltajes[np.isfinite(voltajes)]

        n = len(voltajes)

        if n < 2:
            continue

        # Separaciones consecutivas
        separaciones = np.diff(voltajes)

        media = np.mean(separaciones)

        desviacion = (
            np.std(separaciones, ddof=1)
            if len(separaciones) >= 2 else np.nan
        )

        # Ajuste lineal de la posición de los mínimos
        if n >= 3:

            orden = np.arange(n)

            ajuste = linregress(orden, voltajes)

            energia_ajuste = ajuste.slope
            error_ajuste = ajuste.stderr
            r2 = ajuste.rvalue**2

        else:

            energia_ajuste = np.nan
            error_ajuste = np.nan
            r2 = np.nan

        # Discrepancia porcentual respecto a la referencia
        discrepancia = (
            abs(energia_ajuste - ENERGIA_REFERENCIA)
            / ENERGIA_REFERENCIA * 100
            if np.isfinite(energia_ajuste) else np.nan
        )

        resultados.append({
            "Toma": toma,
            "Tramo": tramo,
            "Numero_minimos": n,
            "Numero_separaciones": len(separaciones),
            "Separacion_media_V": media,
            "Desviacion_separaciones_V": desviacion,
            "Energia_ajuste_eV": energia_ajuste,
            "Error_estandar_pendiente_eV": error_ajuste,
            "R2": r2,
            "Discrepancia_porcentual": discrepancia
        })

        for j, delta in enumerate(separaciones):

            separaciones_totales.append({
                "Toma": toma,
                "Tramo": tramo,
                "Separacion": j + 1,
                "Delta_U_V": delta
            })

        print("\n" + "-" * 50)
        print(f"TOMA {toma} — TRAMO {tramo}")
        print("-" * 50)

        print(f"Número de mínimos: {n}")
        print(f"Separación media: {media:.4f} V")

        if np.isfinite(desviacion):
            print(f"Desviación estándar: {desviacion:.4f} V")

        if np.isfinite(energia_ajuste):

            print(f"Energía por ajuste: {energia_ajuste:.4f} eV")
            print(f"Error estándar de pendiente: {error_ajuste:.4f} eV")
            print(f"Coeficiente R²: {r2:.5f}")
            print(f"Discrepancia: {discrepancia:.2f} %")

        else:
            print("No hay suficientes mínimos para un ajuste lineal.")

    tabla_resultados = pd.DataFrame(resultados)
    tabla_separaciones = pd.DataFrame(separaciones_totales)

    return tabla_resultados, tabla_separaciones


# ==================================================
# 2. PROCESAR LAS DOS TOMAS
# ==================================================

resultados_1, separaciones_1 = analizar_estadistica(
    candidatos_1, 1
)

resultados_2, separaciones_2 = analizar_estadistica(
    candidatos_2, 2
)

resultados = pd.concat(
    [resultados_1, resultados_2],
    ignore_index=True
)

separaciones = pd.concat(
    [separaciones_1, separaciones_2],
    ignore_index=True
)


# ==================================================
# 3. EXPORTAR RESULTADOS
# ==================================================

resultados.to_csv(
    "resultados_estadisticos.csv",
    index=False,
    encoding="utf-8-sig"
)

separaciones.to_csv(
    "separaciones_estadisticas.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nTABLA GENERAL DE RESULTADOS")

if not resultados.empty:
    print(resultados.to_string(index=False))
else:
    print("No se obtuvieron resultados estadísticos.")


# ==================================================
# 4. COMPARACIÓN GRÁFICA
# ==================================================

if not resultados.empty:

    graficables = resultados[
        np.isfinite(resultados["Energia_ajuste_eV"])
    ].copy()

    if not graficables.empty:

        etiquetas = [
            f"Toma {int(f.Toma)} - Tramo {int(f.Tramo)}"
            for f in graficables.itertuples()
        ]

        energias = graficables[
            "Energia_ajuste_eV"
        ].to_numpy()

        errores = graficables[
            "Error_estandar_pendiente_eV"
        ].to_numpy()

        posiciones = np.arange(len(energias))

        fig, ax = plt.subplots(figsize=(9, 5.5))

        # Puntos experimentales
        ax.errorbar(
            posiciones,
            energias,
            yerr=errores,
            fmt="o",
            markersize=7,
            capsize=4,
            color="#2864A5",
            label="Energía estimada y error formal"
        )

        # Intervalo de referencia
        ax.axhspan(
            REFERENCIA_MIN,
            REFERENCIA_MAX,
            color="#55A868",
            alpha=0.18,
            label="Intervalo de referencia"
        )

        ax.axhline(
            ENERGIA_REFERENCIA,
            color="#C44E52",
            linestyle="--",
            linewidth=1.5,
            label="Referencia central"
        )

        ax.set_xticks(posiciones)
        ax.set_xticklabels(etiquetas, rotation=20, ha="right")

        ax.set_ylabel("Energía de excitación (eV)")
        ax.set_title("Comparación estadística de las energías")

        ax.grid(alpha=0.2, linestyle=":")
        ax.legend(frameon=False)

        plt.tight_layout()

        plt.savefig(
            "comparacion_estadistica_energias.png",
            bbox_inches="tight"
        )

        plt.show()


print("\nANÁLISIS ESTADÍSTICO FINALIZADO.")

```

```python

import numpy as np
import pandas as pd

# ==================================================
# CONFIGURACIÓN
# ==================================================

LIMITE_MEDIDOR = 12.67

# ==================================================
# FUNCIÓN DE DEPURACIÓN
# ==================================================

def preparar_segmentos(tabla, toma):

    datos = tabla.copy().reset_index(drop=True)

    datos["U2_V"] = pd.to_numeric(
        datos["U2_V"], errors="coerce"
    )

    datos["I_nA"] = pd.to_numeric(
        datos["I_nA"], errors="coerce"
    )

    datos["Registro_original"] = np.arange(
        1, len(datos) + 1
    )

    datos["Estado"] = "Valido"
    datos["Segmento"] = np.nan

    segmento_actual = 0
    voltaje_anterior = None
    segmento_abierto = False

    for i in range(len(datos)):

        u = datos.loc[i, "U2_V"]
        corriente = datos.loc[i, "I_nA"]

        # Datos ausentes o no numéricos
        if not np.isfinite(u) or not np.isfinite(corriente):

            datos.loc[i, "Estado"] = "Dato_incompleto"
            segmento_abierto = False
            voltaje_anterior = None
            continue

        # Lectura en el límite o fuera de escala
        if corriente >= LIMITE_MEDIDOR:

            datos.loc[i, "Estado"] = "Saturado"
            segmento_abierto = False
            voltaje_anterior = None
            continue

        # Nuevo segmento si el voltaje retrocede
        if (
            voltaje_anterior is None
            or u <= voltaje_anterior
            or not segmento_abierto
        ):

            segmento_actual += 1
            segmento_abierto = True

        datos.loc[i, "Segmento"] = segmento_actual

        voltaje_anterior = u

    # ==================================================
    # RESUMEN DE SEGMENTOS
    # ==================================================

    validos = datos[
        datos["Estado"] == "Valido"
    ].copy()

    resumen = []

    for numero, grupo in validos.groupby("Segmento"):

        voltajes = grupo["U2_V"].to_numpy()

        resumen.append({
            "Toma": toma,
            "Segmento": int(numero),
            "Numero_registros": len(grupo),
            "Voltaje_inicial_V": voltajes[0],
            "Voltaje_final_V": voltajes[-1],
            "Rango_voltaje_V": voltajes[-1] - voltajes[0]
        })

    resumen = pd.DataFrame(resumen)

    # ==================================================
    # INFORME DE CALIDAD
    # ==================================================

    print(f"\n========== TOMA {toma} ==========")

    print("Registros totales:", len(datos))

    print(
        "Registros válidos:",
        (datos["Estado"] == "Valido").sum()
    )

    print(
        "Registros saturados:",
        (datos["Estado"] == "Saturado").sum()
    )

    print(
        "Registros incompletos:",
        (datos["Estado"] == "Dato_incompleto").sum()
    )

    print("\nRESUMEN DE SEGMENTOS:")
    print(resumen.to_string(index=False))

    # Exportar resultados
    datos.to_csv(
        f"datos_clasificados_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    resumen.to_csv(
        f"resumen_segmentos_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    return datos, resumen


# ==================================================
# APLICAR A LAS DOS TOMAS
# ==================================================

datos_limpios_1, resumen_segmentos_1 = (
    preparar_segmentos(tabla_1, 1)
)

datos_limpios_2, resumen_segmentos_2 = (
    preparar_segmentos(tabla_2, 2)
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import find_peaks, savgol_filter

# ==================================================
# CONFIGURACIÓN
# ==================================================

VENTANA_SUAVIZADO = 5
ORDEN_POLINOMIO = 2

DISTANCIA_MINIMA = 3

# Umbrales exploratorios de prominencia en nA.
UMBRALES = [0.4, 0.6, 0.8, 1.0]


# ==================================================
# DETECCIÓN DE MÍNIMOS
# ==================================================

def detectar_minimos_mejorado(datos, toma):

    candidatos = []

    validos = datos[
        datos["Estado"] == "Valido"
    ].copy()

    for segmento, grupo in validos.groupby("Segmento"):

        grupo = grupo.sort_values("Registro_original")

        u = grupo["U2_V"].to_numpy(dtype=float)
        corriente = grupo["I_nA"].to_numpy(dtype=float)

        if len(grupo) < 7:
            continue

        # Suavizado únicamente para explorar la forma.
        corriente_suave = savgol_filter(
            corriente,
            window_length=VENTANA_SUAVIZADO,
            polyorder=ORDEN_POLINOMIO
        )

        # Buscar mínimos como máximos de la curva invertida.
        indices, propiedades = find_peaks(
            -corriente_suave,
            prominence=0,
            distance=DISTANCIA_MINIMA
        )

        prominencias = propiedades["prominences"]

        for indice, prominencia in zip(
            indices, prominencias
        ):

            # Guardar los candidatos que superan cada umbral.
            for umbral in UMBRALES:

                if prominencia >= umbral:

                    candidatos.append({
                        "Toma": toma,
                        "Segmento": segmento,
                        "Registro": int(
                            grupo.iloc[indice]["Registro_original"]
                        ),
                        "U2_V": u[indice],
                        "I_original_nA": corriente[indice],
                        "I_suavizada_nA": corriente_suave[indice],
                        "Prominencia_nA": prominencia,
                        "Umbral_nA": umbral
                    })

    columnas = [
        "Toma",
        "Segmento",
        "Registro",
        "U2_V",
        "I_original_nA",
        "I_suavizada_nA",
        "Prominencia_nA",
        "Umbral_nA"
    ]

    resultado = pd.DataFrame(
        candidatos,
        columns=columnas
    )

    if resultado.empty:
        print(f"\nToma {toma}: no se detectaron candidatos.")
        return resultado

    resultado = resultado.sort_values(
        ["Segmento", "Registro", "Umbral_nA"]
    )

    resultado.to_csv(
        f"minimos_mejorados_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== TOMA {toma} ==========")

    print(
        resultado[
            resultado["Umbral_nA"] == 0.8
        ].to_string(index=False)
    )

    return resultado


# ==================================================
# EJECUTAR
# ==================================================

minimos_mejorados_1 = detectar_minimos_mejorado(
    datos_limpios_1, 1
)

minimos_mejorados_2 = detectar_minimos_mejorado(
    datos_limpios_2, 2
)


# ==================================================
# VISUALIZACIÓN
# ==================================================

def graficar_minimos_mejorados(datos, candidatos, toma):

    fig, ax = plt.subplots(figsize=(10, 5))

    for segmento, grupo in datos[
        datos["Estado"] == "Valido"
    ].groupby("Segmento"):

        ax.scatter(
            grupo["U2_V"],
            grupo["I_nA"],
            s=25,
            alpha=0.65,
            label=f"Segmento {int(segmento)}"
        )

    if not candidatos.empty:

        seleccionados = candidatos[
            np.isclose(candidatos["Umbral_nA"], 0.8)
        ]

        ax.scatter(
            seleccionados["U2_V"],
            seleccionados["I_original_nA"],
            marker="v",
            s=100,
            color="red",
            edgecolors="black",
            linewidths=0.5,
            label="Mínimos candidatos",
            zorder=5
        )

    ax.set_title(
        f"Oscilaciones de corriente - Toma {toma}"
    )

    ax.set_xlabel("Voltaje acelerador $U_2$ (V)")
    ax.set_ylabel("Corriente colectora (nA)")

    ax.grid(alpha=0.25)
    ax.legend(frameon=False)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()

    plt.savefig(
        f"oscilaciones_minimos_toma_{toma}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


graficar_minimos_mejorados(
    datos_limpios_1, minimos_mejorados_1, 1
)

graficar_minimos_mejorados(
    datos_limpios_2, minimos_mejorados_2, 2
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import find_peaks, savgol_filter

# ==================================================
# CONFIGURACIÓN
# ==================================================

VENTANA_SUAVIZADO = 5
ORDEN_POLINOMIO = 2

DISTANCIA_MINIMA = 3

UMBRALES = [0.4, 0.6, 0.8, 1.0]

# ==================================================
# FUNCIÓN PARA DETECTAR SECUENCIAS CRECIENTES
# ==================================================

def obtener_secuencias(grupo):

    u = grupo["U2_V"].to_numpy(dtype=float)

    # Cada retroceso o repetición inicia otra secuencia.
    cortes = np.where(np.diff(u) <= 0)[0] + 1

    return np.split(np.arange(len(grupo)), cortes)


# ==================================================
# DETECCIÓN DE MÍNIMOS
# ==================================================

def detectar_minimos_mejorado(datos, toma):

    candidatos = []

    validos = datos[
        (datos["Estado"] == "Valido")
    ].copy()

    for segmento, grupo in validos.groupby("Segmento"):

        grupo = grupo.sort_values(
            "Registro_original"
        ).reset_index(drop=True)

        secuencias = obtener_secuencias(grupo)

        for numero_secuencia, indices in enumerate(
            secuencias, start=1
        ):

            secuencia = grupo.iloc[indices].copy()

            u = secuencia["U2_V"].to_numpy(dtype=float)

            corriente = secuencia[
                "I_nA"
            ].to_numpy(dtype=float)

            # Se necesitan suficientes puntos para suavizar.
            if len(secuencia) < VENTANA_SUAVIZADO:
                continue

            # Suavizado local.
            corriente_suave = savgol_filter(
                corriente,
                window_length=VENTANA_SUAVIZADO,
                polyorder=ORDEN_POLINOMIO
            )

            # Buscar mínimos locales.
            indices_minimos, propiedades = find_peaks(
                -corriente_suave,
                prominence=0,
                distance=DISTANCIA_MINIMA
            )

            prominencias = propiedades["prominences"]

            for indice, prominencia in zip(
                indices_minimos, prominencias
            ):

                for umbral in UMBRALES:

                    if prominencia >= umbral:

                        candidatos.append({
                            "Toma": toma,
                            "Segmento": segmento,
                            "Secuencia": numero_secuencia,
                            "Registro": int(
                                secuencia.iloc[indice][
                                    "Registro_original"
                                ]
                            ),
                            "U2_V": u[indice],
                            "I_original_nA": corriente[indice],
                            "I_suavizada_nA": corriente_suave[indice],
                            "Prominencia_nA": prominencia,
                            "Umbral_nA": umbral
                        })

    columnas = [
        "Toma",
        "Segmento",
        "Secuencia",
        "Registro",
        "U2_V",
        "I_original_nA",
        "I_suavizada_nA",
        "Prominencia_nA",
        "Umbral_nA"
    ]

    resultado = pd.DataFrame(
        candidatos,
        columns=columnas
    )

    if resultado.empty:

        print(
            f"\nToma {toma}: no se encontraron mínimos candidatos."
        )

        return resultado

    resultado = resultado.sort_values(
        ["Secuencia", "Registro", "Umbral_nA"]
    )

    resultado.to_csv(
        f"minimos_mejorados_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== TOMA {toma} ==========")

    print(
        resultado[
            np.isclose(resultado["Umbral_nA"], 0.8)
        ].to_string(index=False)
    )

    return resultado


# ==================================================
# EJECUTAR DETECCIÓN
# ==================================================

minimos_mejorados_1 = detectar_minimos_mejorado(
    datos_limpios_1, 1
)

minimos_mejorados_2 = detectar_minimos_mejorado(
    datos_limpios_2, 2
)


# ==================================================
# VISUALIZACIÓN
# ==================================================

def graficar_minimos_mejorados(datos, candidatos, toma):

    fig, ax = plt.subplots(figsize=(10, 5.5))

    validos = datos[
        datos["Estado"] == "Valido"
    ].copy()

    # Mostrar los datos experimentales sin unirlos
    # a través de retrocesos.
    ax.scatter(
        validos["U2_V"],
        validos["I_nA"],
        color="#2864A5",
        s=30,
        alpha=0.75,
        label="Datos experimentales",
        zorder=2
    )

    # Conectar únicamente puntos consecutivos
    # cuyo voltaje sea estrictamente creciente.
    for _, grupo in validos.groupby("Segmento"):

        grupo = grupo.sort_values(
            "Registro_original"
        ).reset_index(drop=True)

        secuencias = obtener_secuencias(grupo)

        for indices in secuencias:

            secuencia = grupo.iloc[indices]

            if len(secuencia) >= 2:

                ax.plot(
                    secuencia["U2_V"],
                    secuencia["I_nA"],
                    color="#2864A5",
                    alpha=0.3,
                    linewidth=1
                )

    # Marcar candidatos con prominencia de 0.8 nA.
    if not candidatos.empty:

        seleccionados = candidatos[
            np.isclose(
                candidatos["Umbral_nA"], 0.8
            )
        ]

        if not seleccionados.empty:

            ax.scatter(
                seleccionados["U2_V"],
                seleccionados["I_original_nA"],
                marker="v",
                s=110,
                color="#D62728",
                edgecolors="black",
                linewidths=0.6,
                label="Mínimos candidatos",
                zorder=5
            )

            for _, fila in seleccionados.iterrows():

                ax.annotate(
                    f'{fila["U2_V"]:.1f} V',
                    (
                        fila["U2_V"],
                        fila["I_original_nA"]
                    ),
                    xytext=(0, -16),
                    textcoords="offset points",
                    ha="center",
                    fontsize=8
                )

    ax.set_title(
        f"Detección de mínimos candidatos — Toma {toma}"
    )

    ax.set_xlabel("Voltaje acelerador $U_2$ (V)")
    ax.set_ylabel("Corriente colectora (nA)")

    ax.grid(alpha=0.22, linestyle=":")
    ax.legend(frameon=False)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()

    plt.savefig(
        f"oscilaciones_minimos_toma_{toma}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ==================================================
# GENERAR GRÁFICAS
# ==================================================

graficar_minimos_mejorados(
    datos_limpios_1,
    minimos_mejorados_1,
    1
)

graficar_minimos_mejorados(
    datos_limpios_2,
    minimos_mejorados_2,
    2
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==================================================
# CONFIGURACIÓN
# ==================================================

ENERGIA_REFERENCIA = 18.7
UMBRAL_PRINCIPAL = 0.8

# ==================================================
# ANÁLISIS DE SEPARACIONES
# ==================================================

def analizar_periodicidad(candidatos, toma):

    resultados = []
    separaciones = []

    if candidatos.empty:
        print(f"Toma {toma}: no hay candidatos.")
        return pd.DataFrame(), pd.DataFrame()

    # Cada umbral es una selección diferente.
    for (secuencia, umbral), grupo in candidatos.groupby(
        ["Secuencia", "Umbral_nA"]
    ):

        grupo = grupo.sort_values("U2_V")

        voltajes = grupo["U2_V"].to_numpy(dtype=float)

        if len(voltajes) < 2:
            continue

        diferencias = np.diff(voltajes)

        # Evitar tratar puntos repetidos como separaciones.
        diferencias = diferencias[diferencias > 0]

        if len(diferencias) == 0:
            continue

        media = np.mean(diferencias)

        desviacion = (
            np.std(diferencias, ddof=1)
            if len(diferencias) >= 2 else np.nan
        )

        # Ajuste lineal únicamente con tres o más mínimos.
        if len(voltajes) >= 3:

            orden = np.arange(len(voltajes))

            pendiente, intercepto = np.polyfit(
                orden, voltajes, 1
            )

            prediccion = intercepto + pendiente * orden

            ss_res = np.sum((voltajes - prediccion)**2)

            ss_tot = np.sum(
                (voltajes - np.mean(voltajes))**2
            )

            r2 = (
                1 - ss_res / ss_tot
                if ss_tot > 0 else np.nan
            )

        else:

            pendiente = np.nan
            intercepto = np.nan
            r2 = np.nan

        resultados.append({
            "Toma": toma,
            "Secuencia": secuencia,
            "Umbral_nA": umbral,
            "Numero_minimos": len(voltajes),
            "Separacion_media_V": media,
            "Desviacion_V": desviacion,
            "Energia_estimada_eV": media,
            "Pendiente_ajuste_V": pendiente,
            "R2": r2
        })

        # Guardar cada separación individual.
        for i, diferencia in enumerate(np.diff(voltajes)):

            separaciones.append({
                "Toma": toma,
                "Secuencia": secuencia,
                "Umbral_nA": umbral,
                "U2_inicial_V": voltajes[i],
                "U2_final_V": voltajes[i + 1],
                "Separacion_V": diferencia
            })

    tabla_resultados = pd.DataFrame(resultados)
    tabla_separaciones = pd.DataFrame(separaciones)

    # ==================================================
    # EXPORTAR
    # ==================================================

    tabla_resultados.to_csv(
        f"validacion_periodicidad_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    tabla_separaciones.to_csv(
        f"separaciones_candidatos_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== RESULTADOS TOMA {toma} ==========")

    if tabla_resultados.empty:
        print("No existen separaciones calculables.")
    else:
        print(tabla_resultados.to_string(index=False))

    return tabla_resultados, tabla_separaciones


# ==================================================
# EJECUTAR
# ==================================================

periodicidad_1, separaciones_1 = analizar_periodicidad(
    minimos_mejorados_1, 1
)

periodicidad_2, separaciones_2 = analizar_periodicidad(
    minimos_mejorados_2, 2
)


# ==================================================
# GRÁFICA COMPARATIVA
# ==================================================

fig, ax = plt.subplots(figsize=(9, 5))

for toma, tabla in [
    (1, separaciones_1),
    (2, separaciones_2)
]:

    if tabla.empty:
        continue

    seleccion = tabla[
        np.isclose(
            tabla["Umbral_nA"],
            UMBRAL_PRINCIPAL
        )
    ]

    if not seleccion.empty:

        ax.scatter(
            np.full(len(seleccion), toma),
            seleccion["Separacion_V"],
            s=80,
            label=f"Toma {toma}"
        )

ax.axhline(
    ENERGIA_REFERENCIA,
    color="red",
    linestyle="--",
    label="Referencia: 18.7 V"
)

ax.set_xticks([1, 2])
ax.set_xticklabels(["Primera toma", "Segunda toma"])

ax.set_ylabel("Separación entre mínimos (V)")
ax.set_title("Comparación de separaciones experimentales")

ax.grid(alpha=0.25)
ax.legend(frameon=False)

plt.tight_layout()

plt.savefig(
    "comparacion_separaciones_franck_hertz.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

import numpy as np
import pandas as pd

# ==================================================
# CONFIGURACIÓN
# ==================================================

VENTANA = 2
TOLERANCIA = 0.02

# ==================================================
# VALIDACIÓN DE CADA CANDIDATO
# ==================================================

def validar_minimos_originales(tabla, candidatos, toma):

    resultados = []

    if candidatos.empty:
        print(f"No hay candidatos para la toma {toma}.")
        return pd.DataFrame()

    datos = tabla.copy()

    datos["Registro_original"] = pd.to_numeric(
        datos["Registro_original"], errors="coerce"
    )

    datos["U2_V"] = pd.to_numeric(
        datos["U2_V"], errors="coerce"
    )

    datos["I_nA"] = pd.to_numeric(
        datos["I_nA"], errors="coerce"
    )

    seleccion = candidatos[
        np.isclose(candidatos["Umbral_nA"], 0.8)
    ]

    for _, candidato in seleccion.iterrows():

        registro = int(candidato["Registro"])

        # Vecindad alrededor del candidato
        vecinos = datos[
            (datos["Registro_original"] >= registro - VENTANA)
            & (datos["Registro_original"] <= registro + VENTANA)
        ].copy()

        vecinos = vecinos.sort_values("Registro_original")

        # Comprobar que los registros son consecutivos
        registros = vecinos["Registro_original"].to_numpy()

        indices_consecutivos = (
            len(registros) == 2 * VENTANA + 1
            and np.all(np.diff(registros) == 1)
        )

        if not indices_consecutivos:
            estado = "Vecindad incompleta"
            es_minimo = False
            es_meseta = False

        else:

            corriente = vecinos["I_nA"].to_numpy(dtype=float)

            centro = VENTANA
            valor = corriente[centro]

            izquierda = corriente[:centro]
            derecha = corriente[centro + 1:]

            vecinos_validos = (
                np.all(np.isfinite(corriente))
                and np.all(
                    vecinos["Estado"].to_numpy() == "Valido"
                )
            )

            if not vecinos_validos:

                estado = "Vecindad con datos invalidos"
                es_minimo = False
                es_meseta = False

            else:

                es_minimo = (
                    valor <= np.min(izquierda)
                    and valor <= np.min(derecha)
                )

                # Comprobar si hay una meseta alrededor del candidato
                diferencias = np.abs(corriente - valor)

                es_meseta = (
                    np.sum(diferencias <= TOLERANCIA) > 1
                )

                if es_minimo and es_meseta:
                    estado = "Minimo en meseta"

                elif es_minimo:
                    estado = "Minimo local confirmado"

                else:
                    estado = "No es minimo en datos originales"

        resultados.append({
            "Toma": toma,
            "Registro": registro,
            "U2_V": candidato["U2_V"],
            "Corriente_original_nA": candidato["I_original_nA"],
            "Corriente_suavizada_nA": candidato["I_suavizada_nA"],
            "Prominencia_nA": candidato["Prominencia_nA"],
            "Validacion": estado,
            "Es_minimo_local": es_minimo,
            "Es_meseta": es_meseta
        })

    resultado = pd.DataFrame(resultados)

    resultado.to_csv(
        f"validacion_minimos_originales_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== VALIDACIÓN TOMA {toma} ==========")

    print(resultado.to_string(index=False))

    return resultado


# ==================================================
# EJECUCIÓN
# ==================================================

validacion_1 = validar_minimos_originales(
    datos_limpios_1,
    minimos_mejorados_1,
    1
)

validacion_2 = validar_minimos_originales(
    datos_limpios_2,
    minimos_mejorados_2,
    2
)
```

```python

import numpy as np
import pandas as pd

# ==================================================
# CONFIGURACIÓN
# ==================================================

UMBRAL_PROMINENCIA = 0.8
LIMITE_FONDO = 0.1

# ==================================================
# CARACTERIZACIÓN DE CANDIDATOS
# ==================================================

def caracterizar_valles(tabla, valles, toma):

    if valles.empty:
        print(f"No hay valles para la toma {toma}.")
        return pd.DataFrame(), pd.DataFrame()

    # Eliminar duplicados por umbral.
    unicos = (
        valles.sort_values("Prominencia_nA")
        .drop_duplicates(
            subset=["Toma", "Secuencia", "Registro"],
            keep="last"
        )
        .copy()
    )

    unicos = unicos.sort_values(
        ["Secuencia", "U2_V"]
    )

    resultados = []

    for _, candidato in unicos.iterrows():

        registro = int(candidato["Registro"])

        vecinos = tabla[
            (tabla["Registro_original"] >= registro - 3)
            & (tabla["Registro_original"] <= registro + 3)
            & (tabla["Estado"] == "Valido")
        ].sort_values("Registro_original")

        izquierda = vecinos[
            vecinos["Registro_original"] < registro
        ]["I_nA"]

        derecha = vecinos[
            vecinos["Registro_original"] > registro
        ]["I_nA"]

        corriente = candidato["I_nA"]

        max_izq = (
            izquierda.max() if len(izquierda) else np.nan
        )

        max_der = (
            derecha.max() if len(derecha) else np.nan
        )

        contraste_izq = (
            max_izq - corriente
            if np.isfinite(max_izq) else np.nan
        )

        contraste_der = (
            max_der - corriente
            if np.isfinite(max_der) else np.nan
        )

        # Clasificación exploratoria.
        if corriente <= LIMITE_FONDO:
            clasificacion = "Fondo instrumental probable"

        elif candidato["Prominencia_nA"] < UMBRAL_PROMINENCIA:
            clasificacion = "Valle poco prominente"

        else:
            clasificacion = "Valle candidato"

        resultados.append({
            "Toma": toma,
            "Secuencia": candidato["Secuencia"],
            "Registro": registro,
            "U2_V": candidato["U2_V"],
            "I_nA": corriente,
            "Prominencia_nA": candidato["Prominencia_nA"],
            "Contraste_izquierdo_nA": contraste_izq,
            "Contraste_derecho_nA": contraste_der,
            "Clasificacion": clasificacion
        })

    caracterizacion = pd.DataFrame(resultados)

    # ==================================================
    # SEPARACIONES DENTRO DE CADA SECUENCIA
    # ==================================================

    separaciones = []

    for secuencia, grupo in caracterizacion.groupby("Secuencia"):

        grupo = grupo.sort_values("U2_V")

        for i in range(len(grupo) - 1):

            actual = grupo.iloc[i]
            siguiente = grupo.iloc[i + 1]

            separaciones.append({
                "Toma": toma,
                "Secuencia": secuencia,
                "U2_inicial_V": actual["U2_V"],
                "U2_final_V": siguiente["U2_V"],
                "Separacion_V": (
                    siguiente["U2_V"] - actual["U2_V"]
                ),
                "Clasificacion_inicial": actual["Clasificacion"],
                "Clasificacion_final": siguiente["Clasificacion"]
            })

    tabla_separaciones = pd.DataFrame(separaciones)

    # ==================================================
    # EXPORTACIÓN
    # ==================================================

    caracterizacion.to_csv(
        f"caracterizacion_valles_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    tabla_separaciones.to_csv(
        f"separaciones_depuradas_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== CARACTERIZACIÓN TOMA {toma} ==========")
    print(caracterizacion.to_string(index=False))

    print(f"\n========== SEPARACIONES TOMA {toma} ==========")

    if tabla_separaciones.empty:
        print("No hay separaciones calculables.")
    else:
        print(tabla_separaciones.to_string(index=False))

    return caracterizacion, tabla_separaciones


# ==================================================
# EJECUCIÓN
# ==================================================

caracterizacion_1, separaciones_depuradas_1 = (
    caracterizar_valles(datos_limpios_1, valles_1, 1)
)

caracterizacion_2, separaciones_depuradas_2 = (
    caracterizar_valles(datos_limpios_2, valles_2, 2)
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import peak_prominences

# ==================================================
# DETECCIÓN EXHAUSTIVA DE MÍNIMOS
# ==================================================

def detectar_minimos_locales(tabla, toma):

    datos = tabla.sort_values("Registro_original").copy()

    bloques = []
    actual = []

    # Separar secuencias crecientes y datos consecutivos
    for _, fila in datos.iterrows():

        valido = (
            fila["Estado"] == "Valido"
            and np.isfinite(fila["U2_V"])
            and np.isfinite(fila["I_nA"])
        )

        if not valido:

            if len(actual) >= 3:
                bloques.append(pd.DataFrame(actual))

            actual = []
            continue

        if actual:

            anterior = actual[-1]

            salto = (
                fila["Registro_original"]
                != anterior["Registro_original"] + 1
            )

            retroceso = fila["U2_V"] <= anterior["U2_V"]

            if salto or retroceso:

                if len(actual) >= 3:
                    bloques.append(pd.DataFrame(actual))

                actual = []

        actual.append(fila.to_dict())

    if len(actual) >= 3:
        bloques.append(pd.DataFrame(actual))

    resultados = []

    # Buscar mínimos sin imponer prominencia ni distancia
    for numero, bloque in enumerate(bloques, start=1):

        bloque = bloque.reset_index(drop=True)

        corriente = bloque["I_nA"].to_numpy(dtype=float)
        voltaje = bloque["U2_V"].to_numpy(dtype=float)

        # Mínimo local: menor o igual que ambos vecinos,
        # con al menos una desigualdad estricta.
        indices = []

        for i in range(1, len(corriente) - 1):

            es_minimo = (
                corriente[i] <= corriente[i - 1]
                and corriente[i] <= corriente[i + 1]
                and (
                    corriente[i] < corriente[i - 1]
                    or corriente[i] < corriente[i + 1]
                )
            )

            if es_minimo:
                indices.append(i)

        if not indices:
            continue

        prominencias = peak_prominences(
            -corriente,
            indices
        )[0]

        for indice, prominencia in zip(indices, prominencias):

            resultados.append({
                "Toma": toma,
                "Secuencia": numero,
                "Registro": int(
                    bloque.loc[indice, "Registro_original"]
                ),
                "U2_V": voltaje[indice],
                "I_nA": corriente[indice],
                "Prominencia_nA": prominencia,
                "Delta_I_izquierda_nA": (
                    corriente[indice - 1] - corriente[indice]
                ),
                "Delta_I_derecha_nA": (
                    corriente[indice + 1] - corriente[indice]
                )
            })

    resultado = pd.DataFrame(resultados)

    if resultado.empty:
        print(f"No hay mínimos locales en la toma {toma}.")
        return resultado

    resultado.to_csv(
        f"minimos_locales_exhaustivos_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\n========== MÍNIMOS LOCALES TOMA {toma} ==========")
    print(resultado.to_string(index=False))

    return resultado


# ==================================================
# EJECUCIÓN
# ==================================================

minimos_locales_1 = detectar_minimos_locales(
    datos_limpios_1, 1
)

minimos_locales_2 = detectar_minimos_locales(
    datos_limpios_2, 2
)


# ==================================================
# GRÁFICA DE TODOS LOS MÍNIMOS
# ==================================================

def graficar_minimos_locales(tabla, minimos, toma):

    fig, ax = plt.subplots(figsize=(11, 5.5))

    ax.scatter(
        tabla["U2_V"],
        tabla["I_nA"],
        s=35,
        alpha=0.8,
        label="Mediciones originales"
    )

    if not minimos.empty:

        ax.scatter(
            minimos["U2_V"],
            minimos["I_nA"],
            marker="X",
            s=110,
            color="red",
            edgecolor="black",
            label="Mínimos locales",
            zorder=5
        )

        for _, fila in minimos.iterrows():

            ax.annotate(
                f'{fila["U2_V"]:.1f} V',
                (fila["U2_V"], fila["I_nA"]),
                xytext=(5, 8),
                textcoords="offset points",
                fontsize=8
            )

    ax.set_xlabel(r"Voltaje $U_2$ (V)")
    ax.set_ylabel("Corriente (nA)")
    ax.set_title(f"Mínimos locales de la toma {toma}")

    ax.grid(alpha=0.25)
    ax.legend(frameon=False)

    plt.tight_layout()

    plt.savefig(
        f"minimos_locales_exhaustivos_toma_{toma}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


graficar_minimos_locales(
    datos_limpios_1, minimos_locales_1, 1
)

graficar_minimos_locales(
    datos_limpios_2, minimos_locales_2, 2
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import lombscargle

# ============================================================
# CONFIGURACIÓN
# ============================================================

ENERGIA_REFERENCIA = 18.7  # eV
GRADO_TENDENCIA = 2
NUM_PERIODOS = 2000

PERIODO_MINIMO = 5.0
MINIMO_REGISTROS = 12


# ============================================================
# 1. SEPARAR SECUENCIAS EXPERIMENTALES
# ============================================================

def obtener_secuencias(tabla):

    datos = tabla.sort_values(
        "Registro_original"
    ).copy()

    secuencias = []
    actual = []

    for _, fila in datos.iterrows():

        valido = (
            fila["Estado"] == "Valido"
            and np.isfinite(fila["U2_V"])
            and np.isfinite(fila["I_nA"])
        )

        if not valido:

            if len(actual) >= MINIMO_REGISTROS:
                secuencias.append(pd.DataFrame(actual))

            actual = []
            continue

        if actual:

            anterior = actual[-1]

            registro_contiguo = (
                fila["Registro_original"]
                == anterior["Registro_original"] + 1
            )

            voltaje_creciente = (
                fila["U2_V"] > anterior["U2_V"]
            )

            if not registro_contiguo or not voltaje_creciente:

                if len(actual) >= MINIMO_REGISTROS:
                    secuencias.append(pd.DataFrame(actual))

                actual = []

        actual.append(fila.to_dict())

    if len(actual) >= MINIMO_REGISTROS:
        secuencias.append(pd.DataFrame(actual))

    return secuencias


# ============================================================
# 2. DETRENDING Y ANÁLISIS ESPECTRAL
# ============================================================

def analizar_oscilaciones(tabla, toma):

    secuencias = obtener_secuencias(tabla)

    resultados = []
    espectros = {}

    for numero, grupo in enumerate(secuencias, start=1):

        grupo = grupo.sort_values("U2_V")

        x = grupo["U2_V"].to_numpy(dtype=float)
        y = grupo["I_nA"].to_numpy(dtype=float)

        amplitud = np.ptp(x)

        # Necesitamos suficiente extensión en voltaje.
        periodo_maximo = amplitud / 1.5

        if periodo_maximo <= PERIODO_MINIMO:
            continue

        # ----------------------------------------------------
        # Ajuste polinómico de la tendencia
        # ----------------------------------------------------

        x_centro = x - np.mean(x)

        coeficientes = np.polyfit(
            x_centro,
            y,
            GRADO_TENDENCIA
        )

        tendencia = np.polyval(
            coeficientes,
            x_centro
        )

        residuos = y - tendencia

        # ----------------------------------------------------
        # Periodograma de Lomb-Scargle
        # ----------------------------------------------------

        periodos = np.linspace(
            PERIODO_MINIMO,
            periodo_maximo,
            NUM_PERIODOS
        )

        frecuencias_angulares = (
            2 * np.pi / periodos
        )

        potencia = lombscargle(
            x_centro,
            residuos - np.mean(residuos),
            frecuencias_angulares,
            normalize=True
        )

        indice_maximo = np.argmax(potencia)

        periodo_dominante = periodos[indice_maximo]
        potencia_maxima = potencia[indice_maximo]

        # Potencia en la región de la referencia
        indice_referencia = np.argmin(
            np.abs(periodos - ENERGIA_REFERENCIA)
        )

        potencia_referencia = potencia[indice_referencia]

        # ----------------------------------------------------
        # Guardar resultados
        # ----------------------------------------------------

        resultados.append({

            "Toma": toma,
            "Secuencia": numero,
            "Numero_registros": len(x),

            "Intervalo_U2_V": amplitud,

            "Periodo_dominante_V": periodo_dominante,

            "Potencia_maxima": potencia_maxima,

            "Potencia_cerca_referencia": potencia_referencia,

            "Periodo_referencia_V": ENERGIA_REFERENCIA,

            "Diferencia_referencia_V": (
                periodo_dominante - ENERGIA_REFERENCIA
            )

        })

        espectros[(toma, numero)] = {
            "U2": x,
            "Corriente": y,
            "Tendencia": tendencia,
            "Residuos": residuos,
            "Periodos": periodos,
            "Potencia": potencia,
            "Periodo_dominante": periodo_dominante
        }

        # ----------------------------------------------------
        # Gráficas de diagnóstico
        # ----------------------------------------------------

        fig, ejes = plt.subplots(
            3, 1,
            figsize=(10, 9)
        )

        # Datos y tendencia
        ejes[0].scatter(
            x, y,
            s=28,
            label="Datos originales"
        )

        ejes[0].plot(
            x, tendencia,
            linewidth=2,
            label="Tendencia polinómica"
        )

        ejes[0].set_ylabel("Corriente (nA)")
        ejes[0].set_title(
            f"Toma {toma} — Secuencia {numero}"
        )

        ejes[0].legend(frameon=False)
        ejes[0].grid(alpha=0.2)

        # Residuos
        ejes[1].scatter(
            x,
            residuos,
            s=28
        )

        ejes[1].axhline(
            0,
            color="gray",
            linestyle="--"
        )

        ejes[1].set_xlabel(r"Voltaje $U_2$ (V)")
        ejes[1].set_ylabel("Residuo (nA)")
        ejes[1].set_title("Señal después de retirar la tendencia")
        ejes[1].grid(alpha=0.2)

        # Periodograma
        ejes[2].plot(
            periodos,
            potencia,
            linewidth=1.5
        )

        ejes[2].axvline(
            ENERGIA_REFERENCIA,
            color="red",
            linestyle="--",
            label="Referencia: 18.7 V"
        )

        ejes[2].axvline(
            periodo_dominante,
            color="green",
            linestyle=":",
            label=f"Máximo: {periodo_dominante:.2f} V"
        )

        ejes[2].set_xlabel("Periodo candidato (V)")
        ejes[2].set_ylabel("Potencia normalizada")
        ejes[2].set_title("Periodograma de Lomb–Scargle")

        ejes[2].legend(frameon=False)
        ejes[2].grid(alpha=0.2)

        plt.tight_layout()

        plt.savefig(
            f"analisis_oscilaciones_toma_{toma}_secuencia_{numero}.png",
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()

    resultado = pd.DataFrame(resultados)

    if resultado.empty:

        print(
            f"No hay secuencias suficientemente extensas "
            f"para analizar la toma {toma}."
        )

        return resultado, espectros

    resultado.to_csv(
        f"analisis_oscilaciones_toma_{toma}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"\n========== ANÁLISIS DE OSCILACIONES TOMA {toma} =========="
    )

    print(resultado.to_string(index=False))

    return resultado, espectros


# ============================================================
# 3. EJECUCIÓN
# ============================================================

resultado_oscilaciones_1, espectros_1 = analizar_oscilaciones(
    datos_limpios_1, 1
)

resultado_oscilaciones_2, espectros_2 = analizar_oscilaciones(
    datos_limpios_2, 2
)


# ============================================================
# 4. TABLA GLOBAL
# ============================================================

resultado_global = pd.concat(
    [
        resultado_oscilaciones_1,
        resultado_oscilaciones_2
    ],
    ignore_index=True
)

if not resultado_global.empty:

    resultado_global.to_csv(
        "resultado_global_oscilaciones.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("\n========== RESUMEN GLOBAL ==========")
    print(resultado_global.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import lombscargle
from pathlib import Path

# ==========================================================
# CONFIGURACIÓN
# ==========================================================

CARPETA = Path("graficas_informe")
CARPETA.mkdir(exist_ok=True)

# ==========================================================
# RECONSTRUCCIÓN DE SECUENCIAS
# ==========================================================

def reconstruir_secuencias(tabla):

    datos = tabla.copy()

    datos["U2_V"] = pd.to_numeric(
        datos["U2_V"], errors="coerce"
    )

    datos["I_nA"] = pd.to_numeric(
        datos["I_nA"], errors="coerce"
    )

    if "Estado" in datos.columns:
        datos = datos[datos["Estado"] == "Valido"]

    datos = datos.dropna(
        subset=["U2_V", "I_nA"]
    ).copy()

    # Detectar retrocesos en el voltaje.
    retroceso = datos["U2_V"].diff() < 0

    # Cada retroceso inicia una secuencia nueva.
    datos["Secuencia_aux"] = retroceso.cumsum() + 1

    return [
        grupo.sort_values("U2_V").copy()
        for _, grupo in datos.groupby("Secuencia_aux")
    ]


# ==========================================================
# ANÁLISIS Y GRÁFICAS
# ==========================================================

def generar_espectro(datos, nombre, titulo):

    x = datos["U2_V"].to_numpy(dtype=float)
    y = datos["I_nA"].to_numpy(dtype=float)

    # Ordenar datos por voltaje.
    orden = np.argsort(x)
    x = x[orden]
    y = y[orden]

    # Eliminar voltajes repetidos mediante promedio.
    temporal = pd.DataFrame({
        "U2": x,
        "I": y
    })

    temporal = temporal.groupby(
        "U2", as_index=False
    )["I"].mean()

    x = temporal["U2"].to_numpy()
    y = temporal["I"].to_numpy()

    if len(x) < 10:
        print(f"Se omite {titulo}: datos insuficientes.")
        return

    intervalo = np.ptp(x)

    # Tendencia polinómica de segundo grado.
    xc = x - np.mean(x)

    coeficientes = np.polyfit(xc, y, 2)
    tendencia = np.polyval(coeficientes, xc)

    residuos = y - tendencia

    # Lomb-Scargle.
    periodo_minimo = 5.0
    periodo_maximo = intervalo / 1.5

    if periodo_maximo <= periodo_minimo:
        print(f"Se omite {titulo}: intervalo insuficiente.")
        return

    periodos = np.linspace(
        periodo_minimo,
        periodo_maximo,
        2000
    )

    frecuencias = 2 * np.pi / periodos

    potencia = lombscargle(
        xc,
        residuos - np.mean(residuos),
        frecuencias,
        normalize=True
    )

    indice = np.argmax(potencia)

    periodo_dominante = periodos[indice]
    potencia_maxima = potencia[indice]

    # ------------------------------------------------------
    # FIGURA
    # ------------------------------------------------------

    fig, ejes = plt.subplots(
        3, 1,
        figsize=(9, 8)
    )

    fig.suptitle(
        titulo,
        fontsize=13,
        fontweight="bold"
    )

    ejes[0].scatter(
        x, y,
        s=24,
        label="Datos experimentales"
    )

    ejes[0].plot(
        x, tendencia,
        linewidth=1.8,
        label="Tendencia cuadrática"
    )

    ejes[0].set_ylabel("Corriente (nA)")
    ejes[0].set_title("Datos y tendencia")
    ejes[0].legend(frameon=False)
    ejes[0].grid(alpha=0.25)

    ejes[1].scatter(
        x, residuos,
        s=24
    )

    ejes[1].axhline(
        0,
        linestyle="--",
        linewidth=1
    )

    ejes[1].set_ylabel("Residuo (nA)")
    ejes[1].set_title("Residuos")
    ejes[1].grid(alpha=0.25)

    ejes[2].plot(
        periodos,
        potencia,
        linewidth=1.5
    )

    ejes[2].axvline(
        periodo_dominante,
        linestyle="--",
        label=f"Periodo dominante: {periodo_dominante:.2f} V"
    )

    ejes[2].set_xlabel("Periodo (V)")
    ejes[2].set_ylabel("Potencia")
    ejes[2].set_title("Periodograma de Lomb–Scargle")
    ejes[2].legend(frameon=False)
    ejes[2].grid(alpha=0.25)

    fig.tight_layout(rect=[0, 0, 1, 0.96])

    ruta = CARPETA / nombre

    fig.savefig(
        ruta,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    print(f"\n{titulo}")
    print(f"Periodo dominante: {periodo_dominante:.2f} V")
    print(f"Potencia máxima: {potencia_maxima:.4f}")
    print(f"Archivo: {ruta}")


# ==========================================================
# EJECUCIÓN
# ==========================================================

secuencias_1 = reconstruir_secuencias(datos_limpios_1)
secuencias_2 = reconstruir_secuencias(datos_limpios_2)

print("Secuencias detectadas:")
print("Toma 1:", len(secuencias_1))
print("Toma 2:", len(secuencias_2))

# Generar las figuras correspondientes.
if len(secuencias_1) >= 1:
    generar_espectro(
        secuencias_1[0],
        "analisis_oscilaciones_toma_1_secuencia_1.png",
        "Toma 1 - Secuencia 1"
    )

if len(secuencias_2) >= 1:
    generar_espectro(
        secuencias_2[0],
        "analisis_oscilaciones_toma_2_secuencia_1.png",
        "Toma 2 - Secuencia 1"
    )

if len(secuencias_2) >= 2:
    generar_espectro(
        secuencias_2[1],
        "analisis_oscilaciones_toma_2_secuencia_2.png",
        "Toma 2 - Secuencia 2"
    )

print("\nProceso terminado.")
```

```python

import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# ==========================================================
# 1. CONFIGURACIÓN
# ==========================================================

CARPETA = Path("graficas_informe")
CARPETA.mkdir(exist_ok=True)

# Potenciales observados experimentalmente
toma_1 = {
    "Primera franja visible": 8.0,
    "Tres franjas simultáneas": 21.0
}

toma_2 = {
    "Primera franja": 25.7,
    "Segunda franja": 43.1,
    "Tercera franja": 57.2
}

# ==========================================================
# 2. CONSTRUCCIÓN DE LA FIGURA
# ==========================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 9,
    "axes.titlesize": 10,
    "axes.labelsize": 9,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "axes.linewidth": 0.8,
    "savefig.dpi": 400
})

fig, ax = plt.subplots(figsize=(8.5, 4.8))

# Posiciones verticales de las observaciones
posiciones_1 = np.arange(len(toma_1), 0, -1)
posiciones_2 = np.arange(len(toma_2), 0, -1)

# ==========================================================
# 3. PRIMERA TOMA
# ==========================================================

valores_1 = list(toma_1.values())

ax.scatter(
    valores_1,
    posiciones_1,
    s=75,
    marker="o",
    facecolors="white",
    edgecolors="#24577A",
    linewidths=1.8,
    label="Primera toma",
    zorder=3
)

for valor, posicion, etiqueta in zip(
    valores_1,
    posiciones_1,
    toma_1.keys()
):
    ax.annotate(
        f"{valor:.1f} V",
        (valor, posicion),
        xytext=(0, 11),
        textcoords="offset points",
        ha="center",
        fontsize=8
    )

# ==========================================================
# 4. SEGUNDA TOMA
# ==========================================================

valores_2 = list(toma_2.values())

ax.scatter(
    valores_2,
    posiciones_2,
    s=75,
    marker="s",
    facecolors="white",
    edgecolors="#C26A36",
    linewidths=1.8,
    label="Segunda toma",
    zorder=3
)

for valor, posicion in zip(
    valores_2,
    posiciones_2
):
    ax.annotate(
        f"{valor:.1f} V",
        (valor, posicion),
        xytext=(0, 11),
        textcoords="offset points",
        ha="center",
        fontsize=8
    )

# ==========================================================
# 5. FORMATO DE LOS EJES
# ==========================================================

ax.set_yticks([1, 2, 3])

ax.set_yticklabels([
    "Primera franja",
    "Segunda franja",
    "Tercera franja"
])

ax.set_xlim(0, 65)
ax.set_ylim(0.5, 3.5)

ax.set_xlabel(r"Potencial acelerador $U_2$ (V)")

ax.set_title(
    "Potenciales asociados a la observación de franjas luminosas",
    loc="center",
    fontweight="bold",
    pad=14
)

ax.grid(
    axis="x",
    linestyle=":",
    alpha=0.35
)

ax.legend(
    frameon=False,
    loc="lower right"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.tick_params(
    direction="out",
    length=3
)

# ==========================================================
# 6. NOTA EXPERIMENTAL
# ==========================================================

fig.text(
    0.5,
    0.015,
    "La observación de tres franjas simultáneas en la toma 1 "
    "corresponde a un único registro de potencial.",
    ha="center",
    fontsize=8,
    style="italic"
)

# ==========================================================
# 7. EXPORTACIÓN
# ==========================================================

fig.tight_layout(rect=[0.02, 0.07, 0.98, 0.96])

ruta = CARPETA / "franjas_luminosas.png"

fig.savefig(
    ruta,
    dpi=400,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
plt.close(fig)

print("Gráfica generada correctamente.")
print(f"Ubicación: {ruta.resolve()}")
```

```python

import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# FRANJAS LUMINOSAS - EXPERIMENTO DE FRANCK-HERTZ
# ============================================================

CARPETA = Path("graficas_informe")
CARPETA.mkdir(exist_ok=True)

# Potenciales registrados experimentalmente
observaciones = [
    {
        "toma": "Toma 1",
        "descripcion": "Primera franja visible",
        "U2": 8.0,
        "posicion": 4
    },
    {
        "toma": "Toma 1",
        "descripcion": "Tres franjas simultáneas",
        "U2": 21.0,
        "posicion": 3
    },
    {
        "toma": "Toma 2",
        "descripcion": "Primera franja",
        "U2": 25.7,
        "posicion": 2
    },
    {
        "toma": "Toma 2",
        "descripcion": "Segunda franja",
        "U2": 43.1,
        "posicion": 1
    },
    {
        "toma": "Toma 2",
        "descripcion": "Tercera franja",
        "U2": 57.2,
        "posicion": 0
    }
]

# ============================================================
# CONFIGURACIÓN DE LA FIGURA
# ============================================================

fig, ax = plt.subplots(figsize=(10, 5.8))

colores = {
    "Toma 1": "#2878B5",
    "Toma 2": "#D97928"
}

marcadores = {
    "Toma 1": "o",
    "Toma 2": "s"
}

# ============================================================
# REPRESENTACIÓN DE LOS DATOS
# ============================================================

for dato in observaciones:

    ax.scatter(
        dato["U2"],
        dato["posicion"],
        s=100,
        color=colores[dato["toma"]],
        marker=marcadores[dato["toma"]],
        edgecolor="black",
        linewidth=0.7,
        zorder=3,
        label=dato["toma"]
    )

    ax.annotate(
        f'{dato["U2"]:.1f} V',
        xy=(dato["U2"], dato["posicion"]),
        xytext=(0, 12),
        textcoords="offset points",
        ha="center",
        fontsize=9
    )

# ============================================================
# EJES Y ESTÉTICA
# ============================================================

etiquetas = [
    "Toma 2: tercera franja",
    "Toma 2: segunda franja",
    "Toma 2: primera franja",
    "Toma 1: tres franjas simultáneas",
    "Toma 1: primera franja visible"
]

ax.set_yticks(range(5))
ax.set_yticklabels(etiquetas)

ax.set_xlabel(
    r"Potencial acelerador $U_2$ (V)",
    fontsize=11
)

ax.set_ylabel(
    "Observación experimental",
    fontsize=11
)

ax.set_title(
    "Potenciales asociados a las franjas luminosas",
    fontsize=13,
    pad=15
)

ax.set_xlim(0, 65)
ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.35
)

# Leyenda sin entradas duplicadas
handles, labels = ax.get_legend_handles_labels()
unicos = dict(zip(labels, handles))

ax.legend(
    unicos.values(),
    unicos.keys(),
    frameon=False,
    loc="lower right"
)

plt.tight_layout()

# ============================================================
# EXPORTACIÓN
# ============================================================

ruta = CARPETA / "franjas_luminosas.png"

plt.savefig(
    ruta,
    dpi=400,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print(f"Gráfica guardada correctamente en: {ruta}")
```
